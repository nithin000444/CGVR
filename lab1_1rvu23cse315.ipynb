```python
import matplotlib.pyplot as plt


def bresenham_line(x1, y1, x2, y2):
    points = []

    dx = abs(x2 - x1)
    dy = abs(y2 - y1)

    sx = 1 if x2 > x1 else -1
    sy = 1 if y2 > y1 else -1

    x = x1
    y = y1

    # Case 1: |slope| <= 1
    if dx >= dy:
        p = 2 * dy - dx

        for i in range(dx + 1):
            points.append((x, y))

            if p >= 0:
                y = y + sy
                p = p + 2 * (dy - dx)
            else:
                p = p + 2 * dy

            x = x + sx

    # Case 2: |slope| > 1
    else:
        p = 2 * dx - dy

        for i in range(dy + 1):
            points.append((x, y))

            if p >= 0:
                x = x + sx
                p = p + 2 * (dx - dy)
            else:
                p = p + 2 * dx

            y = y + sy

    return points


# Input points
x1 = int(input("Enter x1 of point A: "))
y1 = int(input("Enter y1 of point A: "))

x2 = int(input("Enter x2 of point B: "))
y2 = int(input("Enter y2 of point B: "))


# Calculate points
points = bresenham_line(x1, y1, x2, y2)

print("\nPoints calculated using Bresenham's Algorithm:")
for point in points:
    print(point)


# Separate x and y coordinates
x_points = [p[0] for p in points]
y_points = [p[1] for p in points]


# Draw the line
plt.figure(figsize=(8, 6))

plt.plot(x_points, y_points, 'o-', linewidth=1)

# Mark A and B
plt.scatter([x1, x2], [y1, y2], s=100)

plt.text(x1, y1, "  A", fontsize=12)
plt.text(x2, y2, "  B", fontsize=12)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Bresenham's Line Drawing Algorithm")

plt.grid(True)
plt.xticks(range(min(x1, x2) - 1, max(x1, x2) + 2))
plt.yticks(range(min(y1, y2) - 1, max(y1, y2) + 2))

plt.show()
```

```python
import matplotlib.pyplot as plt


def dda_line(x1, y1, x2, y2):
    points = []

    dx = x2 - x1
    dy = y2 - y1

    # Number of steps
    steps = max(abs(dx), abs(dy))

    # Increment in x and y
    x_increment = dx / steps
    y_increment = dy / steps

    x = x1
    y = y1

    for i in range(steps + 1):
        points.append((round(x), round(y)))

        x = x + x_increment
        y = y + y_increment

    return points


# Input points
x1 = int(input("Enter x1 of point A: "))
y1 = int(input("Enter y1 of point A: "))

x2 = int(input("Enter x2 of point B: "))
y2 = int(input("Enter y2 of point B: "))


# Calculate points
points = dda_line(x1, y1, x2, y2)

print("\nPoints calculated using DDA Algorithm:")
for point in points:
    print(point)


# Separate x and y coordinates
x_points = [p[0] for p in points]
y_points = [p[1] for p in points]


# Draw the line
plt.figure(figsize=(8, 6))

plt.plot(x_points, y_points, 'o-', linewidth=1)

# Mark A and B
plt.scatter([x1, x2], [y1, y2], s=100)

plt.text(x1, y1, "  A", fontsize=12)
plt.text(x2, y2, "  B", fontsize=12)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("DDA Line Drawing Algorithm")

plt.grid(True)

plt.xticks(range(min(x1, x2) - 1, max(x1, x2) + 2))
plt.yticks(range(min(y1, y2) - 1, max(y1, y2) + 2))

plt.show()
```
